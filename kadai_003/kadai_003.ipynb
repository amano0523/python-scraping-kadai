```python
!pip install selenium webdriver-manager

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from getpass import getpass
import time


options = webdriver.ChromeOptions()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

wait = WebDriverWait(driver, 30)

driver.get("https://terakoya.sejuku.net/register")
print("ページ表示OK")

# 右上のログインをクリック
login_link = wait.until(
    EC.element_to_be_clickable((By.XPATH, "//div[text()='ログイン']"))
)
login_link.click()
print("ログイン画面表示")

# ログインフォーム全体を取得
login_form = wait.until(
    EC.visibility_of_element_located(
        (By.CSS_SELECTOR, "div.sc-lgpSej.gsAReM")
    )
)

email_input = login_form.find_element(By.NAME, "email")
password_input = login_form.find_element(By.NAME, "password")

email_input.clear()
email_input.send_keys("amano.keita0523@gmail.com")

password = getpass("パスワードを入力してください: ")
password_input.clear()
password_input.send_keys(password)
password_input.send_keys(Keys.TAB)

time.sleep(2)

# ログインフォーム内のログインボタンだけ取得
login_button = login_form.find_element(
    By.XPATH,
    ".//button[normalize-space()='ログイン']"
)

login_button.click()
print("ログインクリック")

time.sleep(5)

print("現在URL:", driver.current_url)

# プロフィールページへ移動
driver.get("https://terakoya.sejuku.net/account/profile")
print("プロフィールページ表示")

# 編集
edit_button = wait.until(
    EC.element_to_be_clickable((By.XPATH, "//button[text()='編集']"))
)
edit_button.click()
print("編集開始")

# 自己紹介欄
textarea = wait.until(
    EC.visibility_of_element_located((By.TAG_NAME, "textarea"))
)

textarea.clear()
textarea.send_keys(
    "プログラミング学習中です！今はスクレイピングに挑戦しています！"
)
print("自己紹介入力完了")

# 更新
update_button = wait.until(
    EC.element_to_be_clickable((By.XPATH, "//button[text()='更新する']"))
)
update_button.click()

time.sleep(3)

driver.save_screenshot("kadai_003_result.png")

print("更新完了")
print("スクリーンショット保存完了")

driver.quit()
```
