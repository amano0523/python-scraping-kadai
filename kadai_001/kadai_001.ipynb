```python
import requests
from bs4 import BeautifulSoup

url = "https://news.yahoo.co.jp/articles/ab1772be9bc7d7c4042abad3b5fd6ec98b901e81"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"}
)

soup = BeautifulSoup(response.text, "html.parser")

# すべてのpタグを取得
paragraphs = soup.find_all("p")

# 本文を結合
article_text = ""

for p in paragraphs:
    text = p.get_text(strip=True)

    if text:
        article_text += text + "\n"

print(article_text)
```
