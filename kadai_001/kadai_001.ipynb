```python
import requests
from bs4 import BeautifulSoup

url = "https://news.yahoo.co.jp/articles/ab1772be9bc7d7c4042abad3b5fd6ec98b901e81"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"}
)

soup = BeautifulSoup(response.text, "html.parser")

# 記事本文エリアを取得
article = soup.find("article", id="uamods")

# 本文格納用
article_text = ""

if article:
    # article内のpタグを取得
    paragraphs = article.find_all("p")

    for p in paragraphs:
        text = p.get_text(strip=True)

        if text:
            article_text += text + "\n"

print(article_text)
```
