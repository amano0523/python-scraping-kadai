```python
import requests

# APIキーを入力
api_key = input("Google Maps APIキーを入力してください: ")

# 駅名を入力
station_name = input("駅名を入力してください: ")

# Geocoding APIで駅の緯度経度を取得
geocode_url = "https://maps.googleapis.com/maps/api/geocode/json"

geocode_params = {
    "address": station_name,
    "key": api_key
}

geocode_response = requests.get(
    geocode_url,
    params=geocode_params
)

geocode_data = geocode_response.json()

if geocode_data["status"] != "OK":
    print("駅が見つかりませんでした。")
    exit()

location = geocode_data["results"][0]["geometry"]["location"]
lat = location["lat"]
lng = location["lng"]

# Places APIでレストランを検索
places_url = "https://maps.googleapis.com/maps/api/place/nearbysearch/json"

places_params = {
    "location": f"{lat},{lng}",
    "radius": 500,
    "type": "restaurant",
    "key": api_key
}

places_response = requests.get(
    places_url,
    params=places_params
)

places_data = places_response.json()

print("\n【検索結果】")

for restaurant in places_data["results"]:
    name = restaurant.get("name", "情報なし")
    rating = restaurant.get("rating", "評価なし")
    address = restaurant.get("vicinity", "住所なし")

    print("--------------------")
    print("レストラン名:", name)
    print("評価:", rating)
    print("住所:", address)
```
